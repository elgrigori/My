## Dataset and Preprocessing

```python
nltk.download('reuters')
nltk.download('punkt')
nltk.download('punkt_tab')

# Load all sentence-tokenized Reuters corpus sentences.
sentences = reuters.sents()

print("=" * 70)
print("REUTERS DATASET SUMMARY")
print("=" * 70)
print(f"Number of raw sentences: {len(sentences):,}")
print(f"Example raw sentence: {list(sentences[0])}")
```

### Preprocessing Details

Raw corpus text contains capitalization, punctuation, rare words, and formatting differences that increase sparsity. The preprocessing steps below normalize the corpus before n-gram counting.

**Lowercasing** maps forms such as `The`, `the`, and `THE` to the same token.

**Punctuation cleaning** removes punctuation-only tokens and strips punctuation characters while preserving apostrophes inside contractions such as `don't`.

**Vocabulary cutoff** keeps only words that occur at least `MIN_COUNT` times in the training set. This reduces the number of extremely sparse word types.

**UNK replacement** maps every word outside the vocabulary to `*UNK*`, so the model can assign probability to unseen or rare words.

**Train/dev/test split** uses the training set for counts, the development set for choosing alpha, and the test set only for final evaluation.

```python
import re


def clean_token(token):
    """Lowercase a token and remove punctuation except internal apostrophes."""
    token = token.lower()
    token = re.sub(r"[^a-z0-9']", "", token)
    return token if token else None


def preprocess_sentence(sentence):
    """Apply the token cleaning pipeline to one sentence."""
    cleaned_tokens = []
    for token in sentence:
        cleaned = clean_token(str(token))
        if cleaned is not None:
            cleaned_tokens.append(cleaned)
    return cleaned_tokens


processed_sentences = [preprocess_sentence(sent) for sent in sentences]
processed_sentences = [sent for sent in processed_sentences if sent]

print("=" * 70)
print("PREPROCESSING SUMMARY")
print("=" * 70)
print(f"Sentences after preprocessing: {len(processed_sentences):,}")
print("First 3 processed sentences:")
for sent in processed_sentences[:3]:
    print(" ", sent)
```

```python
# Quick validation that punctuation-only tokens were removed while useful tokens remain.
all_processed_tokens = {token for sent in processed_sentences for token in sent}
punctuation_check = ['.', ',', '!', '?', ';', ':', '(', ')', '"', "'", '``', "''", '--', '-', '/']
found_punctuation = [punct for punct in punctuation_check if punct in all_processed_tokens]
contractions_found = sorted(token for token in all_processed_tokens if "'" in token)[:5]
numbers_found = sorted(token for token in all_processed_tokens if any(c.isdigit() for c in token))[:5]

print("=" * 70)
print("PREPROCESSING VALIDATION")
print("=" * 70)
print(f"Standalone punctuation found: {found_punctuation if found_punctuation else 'none'}")
print(f"Contraction examples: {contractions_found}")
print(f"Number examples: {numbers_found}")
print(f"Unique tokens after cleaning: {len(all_processed_tokens):,}")
```

### Train / Development / Test Split

The corpus is shuffled with a fixed random seed for reproducibility and split into:

- **80% train**: estimate n-gram counts
- **10% development**: choose the smoothing parameter alpha
- **10% test**: final held-out evaluation

```python
random.seed(42)

sentences_shuffled = processed_sentences.copy()
random.shuffle(sentences_shuffled)

n = len(sentences_shuffled)
train_end = int(0.80 * n)
dev_end = int(0.90 * n)

train_sents = sentences_shuffled[:train_end]
dev_sents = sentences_shuffled[train_end:dev_end]
test_sents = sentences_shuffled[dev_end:]

print("=" * 70)
print("DATA SPLIT")
print("=" * 70)
print(f"All sentences:   {len(processed_sentences):,}")
print(f"Train sentences: {len(train_sents):,}")
print(f"Dev sentences:   {len(dev_sents):,}")
print(f"Test sentences:  {len(test_sents):,}")
```

### Vocabulary and UNK Replacement

The vocabulary is built **only from the training set**. Words with fewer than `MIN_COUNT = 10` training occurrences are replaced by `*UNK*` in train, development, and test data.

The prediction vocabulary includes `*UNK*` and `*end*`, but it does not include START tokens because START tokens are conditioning context, not words the model should predict.

```python
MIN_COUNT = 10
UNK = "*UNK*"
END = "*end*"
START = "*start*"
START1 = "*start1*"
START2 = "*start2*"

word_counts = Counter(token for sent in train_sents for token in sent)

# Prediction vocabulary: frequent training words plus UNK and END.
# START tokens are context markers and are not predicted.
vocabulary = {word for word, count in word_counts.items() if count >= MIN_COUNT}
vocabulary.add(UNK)
vocabulary.add(END)

print("=" * 70)
print("CORPUS AND VOCABULARY STATISTICS")
print("=" * 70)
print("Corpus: Reuters")
print(f"Number of sentences: {len(processed_sentences):,}")
print(f"Training sentences:  {len(train_sents):,}")
print(f"Dev sentences:       {len(dev_sents):,}")
print(f"Test sentences:      {len(test_sents):,}")
print(f"Minimum count cutoff: {MIN_COUNT}")
print(f"Vocabulary size:      {len(vocabulary):,}")
print(f"Total train tokens:   {sum(word_counts.values()):,}")
print("Sample vocabulary items:")
print(sorted(vocabulary)[:30])
```

## Smoothing

### Laplace / Add-alpha Smoothing

A raw n-gram model assigns probability 0 to any n-gram that never appeared in the training set. This is a serious problem because one unseen phrase would make the probability of a whole test sentence equal to 0.

Add-alpha smoothing fixes this by adding a small constant `alpha` to every possible next-word count.

For bigrams:

$$P(w_i \mid w_{i-1}) = \frac{count(w_{i-1}, w_i) + \alpha}{count(w_{i-1}) + \alpha |V|}$$

For trigrams:

$$P(w_i \mid w_{i-2}, w_{i-1}) = \frac{count(w_{i-2}, w_{i-1}, w_i) + \alpha}{count(w_{i-2}, w_{i-1}) + \alpha |V|}$$

`alpha = 1.0` is standard Laplace smoothing. Smaller values such as `0.1` or `0.5` smooth less aggressively and often work better for language modeling, so we tune alpha on the development set.

```python
def replace_oov(sentence, vocabulary):
    """Replace tokens outside the vocabulary with UNK."""
    return [token if token in vocabulary else UNK for token in sentence]


train_sents_unk = [replace_oov(sent, vocabulary) for sent in train_sents]
dev_sents_unk = [replace_oov(sent, vocabulary) for sent in dev_sents]
test_sents_unk = [replace_oov(sent, vocabulary) for sent in test_sents]

print("=" * 70)
print("UNK REPLACEMENT SUMMARY")
print("=" * 70)
print("Example before OOV replacement:")
print(train_sents[0])
print("Example after OOV replacement:")
print(train_sents_unk[0])
```

## Bigram Language Model

The bigram model estimates the probability of each word using only the previous word as context:

$$P(w_i \mid w_{i-1})$$

Each training sentence receives one `*start*` token at the beginning and one `*end*` token at the end. The model counts `(previous_word, current_word)` pairs and applies add-alpha smoothing when computing probabilities.

```python
class BigramLanguageModel:
    def __init__(self, vocabulary, alpha=1.0):
        self.vocabulary = set(vocabulary)
        self.vocab_size = len(self.vocabulary)
        self.alpha = alpha
        self.bigram_counts = Counter()
        self.context_counts = Counter()

    def train(self, sentences):
        for sent in sentences:
            tokens = [START] + sent + [END]
            for i in range(1, len(tokens)):
                context = tokens[i - 1]
                word = tokens[i]
                self.bigram_counts[(context, word)] += 1
                self.context_counts[context] += 1

    def probability(self, context, word):
        if word not in self.vocabulary:
            word = UNK

        return (
            self.bigram_counts[(context, word)] + self.alpha
        ) / (
            self.context_counts[context] + self.alpha * self.vocab_size
        )


print("BigramLanguageModel class defined")
```

## Trigram Language Model

The trigram model estimates the probability of each word using the two previous words as context:

$$P(w_i \mid w_{i-2}, w_{i-1})$$

Each sentence receives two START tokens, `*start1*` and `*start2*`, so the first real word can still be predicted from a full two-token context.

For smoothing, the denominator uses the **prediction vocabulary** size: regular vocabulary words plus `*UNK*` and `*end*`. START tokens are not predicted words, so they are not added to the prediction vocabulary. They are used only as context markers.

```python
class TrigramLanguageModel:
    def __init__(self, vocabulary, alpha=1.0):
        self.vocabulary = set(vocabulary)
        self.vocab_size = len(self.vocabulary)
        self.alpha = alpha
        self.trigram_counts = Counter()
        self.context_counts = Counter()
        self.start_context_tokens = {START1, START2}

    def _normalize_context_token(self, token):
        """Map non-start context tokens outside the vocabulary to UNK."""
        if token in self.start_context_tokens:
            return token
        return token if token in self.vocabulary else UNK

    def train(self, sentences):
        for sent in sentences:
            tokens = [START1, START2] + sent + [END]
            for i in range(2, len(tokens)):
                context = (tokens[i - 2], tokens[i - 1])
                word = tokens[i]
                self.trigram_counts[(context[0], context[1], word)] += 1
                self.context_counts[context] += 1

    def probability(self, context, word):
        context = (
            self._normalize_context_token(context[0]),
            self._normalize_context_token(context[1]),
        )

        if word not in self.vocabulary:
            word = UNK

        return (
            self.trigram_counts[(context[0], context[1], word)] + self.alpha
        ) / (
            self.context_counts[context] + self.alpha * self.vocab_size
        )


print("TrigramLanguageModel class defined")
```

## Cross-Entropy and Perplexity

### Why Use Log Probabilities?

Language model probabilities are often very small. Multiplying many small probabilities can cause numerical underflow, where the computer rounds the product to zero. Log probabilities avoid this because multiplication becomes addition:

$$\log(p_1 \times p_2 \times ... \times p_N) = \log(p_1) + \log(p_2) + ... + \log(p_N)$$

This notebook uses log base 2, so cross-entropy is measured in bits per token.

### Cross-Entropy

Cross-entropy measures the average surprise of the model on held-out data:

$$H = -\frac{1}{N}\sum_{i=1}^{N}\log_2 P(w_i \mid context_i)$$

Lower cross-entropy means the model assigns higher probability to the actual next words.

### Perplexity

Perplexity is the exponentiated cross-entropy:

$$Perplexity = 2^H$$

It can be interpreted as the model's effective average number of choices at each prediction step. Lower perplexity is better.

### Evaluation Counting Rules

- START tokens are excluded from perplexity computation.
- END tokens are included in `N` because predicting sentence endings is part of the task.
- The entire dev/test set is treated as one evaluation sequence for aggregation: total log probability and total `N` are accumulated over all sentences before computing one cross-entropy and one perplexity score.

```python
def evaluate_bigram(model, sentences):
    total_log_prob = 0.0
    N = 0

    for sent in sentences:
        tokens = [START] + sent + [END]
        # Start at i=1: predict real words and END, but never predict START.
        for i in range(1, len(tokens)):
            context = tokens[i - 1]
            word = tokens[i]
            prob = model.probability(context, word)
            total_log_prob += math.log2(prob)
            N += 1

    cross_entropy = -total_log_prob / N
    perplexity = 2 ** cross_entropy
    return cross_entropy, perplexity, N


def evaluate_trigram(model, sentences):
    total_log_prob = 0.0
    N = 0

    for sent in sentences:
        tokens = [START1, START2] + sent + [END]
        # Start at i=2: predict real words and END, but never predict START tokens.
        for i in range(2, len(tokens)):
            context = (tokens[i - 2], tokens[i - 1])
            word = tokens[i]
            prob = model.probability(context, word)
            total_log_prob += math.log2(prob)
            N += 1

    cross_entropy = -total_log_prob / N
    perplexity = 2 ** cross_entropy
    return cross_entropy, perplexity, N


print("Evaluation functions defined")
```

## Development-set Hyperparameter Tuning

The smoothing parameter `alpha` controls how much probability mass is assigned to unseen n-grams.

We evaluate `alpha` values `[0.1, 0.5, 1.0]` on the development set, select the value with the lowest development perplexity for each model, and then use those selected values for final test evaluation.

```python
alpha_values = [0.1, 0.5, 1.0]

def print_tuning_header():
    print("=" * 100)
    print("DEVELOPMENT-SET ALPHA TUNING")
    print("=" * 100)
    print(f"{'alpha':<8} {'bigram CE':<14} {'bigram PPL':<14} {'trigram CE':<14} {'trigram PPL':<14}")
    print("-" * 100)


print_tuning_header()
tuning_results = {}

for alpha in alpha_values:
    bigram_temp = BigramLanguageModel(vocabulary, alpha=alpha)
    bigram_temp.train(train_sents_unk)

    trigram_temp = TrigramLanguageModel(vocabulary, alpha=alpha)
    trigram_temp.train(train_sents_unk)

    bigram_ce_dev, bigram_perp_dev, _ = evaluate_bigram(bigram_temp, dev_sents_unk)
    trigram_ce_dev, trigram_perp_dev, _ = evaluate_trigram(trigram_temp, dev_sents_unk)

    tuning_results[alpha] = {
        "bigram_ce": bigram_ce_dev,
        "bigram_perp": bigram_perp_dev,
        "trigram_ce": trigram_ce_dev,
        "trigram_perp": trigram_perp_dev,
    }

    print(f"{alpha:<8.1f} {bigram_ce_dev:<14.4f} {bigram_perp_dev:<14.2f} {trigram_ce_dev:<14.4f} {trigram_perp_dev:<14.2f}")

best_alpha_bigram = min(tuning_results, key=lambda a: tuning_results[a]["bigram_perp"])
best_alpha_trigram = min(tuning_results, key=lambda a: tuning_results[a]["trigram_perp"])

print("-" * 100)
print("BEST ALPHA VALUES")
print(f"Bigram:  alpha = {best_alpha_bigram} | dev perplexity = {tuning_results[best_alpha_bigram]['bigram_perp']:.2f}")
print(f"Trigram: alpha = {best_alpha_trigram} | dev perplexity = {tuning_results[best_alpha_trigram]['trigram_perp']:.2f}")
print("=" * 100)

bigram_lm = BigramLanguageModel(vocabulary, alpha=best_alpha_bigram)
bigram_lm.train(train_sents_unk)

trigram_lm = TrigramLanguageModel(vocabulary, alpha=best_alpha_trigram)
trigram_lm.train(train_sents_unk)

print("Final models retrained on the training set with the selected alpha values.")
```

```python
print("=" * 100)
print("FINAL EVALUATION ON TEST SET")
print("=" * 100)
print("Counting rules: START tokens excluded; END tokens included in N.")
print("Aggregation: the full test set is evaluated as one corpus-level sequence.\n")

bigram_ce_test, bigram_perp_test, bigram_N = evaluate_bigram(bigram_lm, test_sents_unk)
trigram_ce_test, trigram_perp_test, trigram_N = evaluate_trigram(trigram_lm, test_sents_unk)

print(f"{'Model':<12} {'Alpha':<8} {'Cross-Entropy':<18} {'Perplexity':<16} {'N tokens':<12}")
print("-" * 100)
print(f"{'Bigram':<12} {best_alpha_bigram:<8.1f} {bigram_ce_test:<18.4f} {bigram_perp_test:<16.2f} {bigram_N:<12}")
print(f"{'Trigram':<12} {best_alpha_trigram:<8.1f} {trigram_ce_test:<18.4f} {trigram_perp_test:<16.2f} {trigram_N:<12}")
print("-" * 100)

print("\nEXPERIMENT SUMMARY")
print(f"Vocabulary size:      {len(vocabulary):,}")
print(f"Training sentences:   {len(train_sents_unk):,}")
print(f"Development sentences:{len(dev_sents_unk):,}")
print(f"Test sentences:       {len(test_sents_unk):,}")
print(f"Best bigram alpha:    {best_alpha_bigram}")
print(f"Best trigram alpha:   {best_alpha_trigram}")

perp_change = ((bigram_perp_test - trigram_perp_test) / bigram_perp_test) * 100
print("\nMODEL COMPARISON")
print(f"Trigram perplexity change vs. bigram: {perp_change:.1f}%")
print("A lower perplexity indicates a better held-out language model.")
```

## Trigram Sparsity Diagnostics

If trigram perplexity is much worse than bigram perplexity, the most likely reason is sparsity rather than an error in the cross-entropy formula. A trigram model conditions on two previous tokens, so it can encounter many contexts in the test set that were never observed in training.

The diagnostics below check:

- how many unique bigrams and trigrams were learned
- how often test trigram contexts were unseen in training
- how often full test trigram events were unseen in training

START tokens are still excluded from `N`, END tokens are still included, and the test set is aggregated as one evaluation sequence.

```python
def trigram_sparsity_diagnostics(model, sentences):
    total_positions = 0
    unseen_contexts = 0
    unseen_events = 0
    unique_unseen_contexts = set()

    for sent in sentences:
        tokens = [START1, START2] + sent + [END]
        for i in range(2, len(tokens)):
            context = (tokens[i - 2], tokens[i - 1])
            context = (
                model._normalize_context_token(context[0]),
                model._normalize_context_token(context[1]),
            )
            word = tokens[i] if tokens[i] in model.vocabulary else UNK
            event = (context[0], context[1], word)

            total_positions += 1

            if model.context_counts.get(context, 0) == 0:
                unseen_contexts += 1
                unique_unseen_contexts.add(context)

            if model.trigram_counts.get(event, 0) == 0:
                unseen_events += 1

    unseen_context_pct = 100 * unseen_contexts / total_positions
    unseen_event_pct = 100 * unseen_events / total_positions

    return {
        "total_positions": total_positions,
        "unseen_contexts": unseen_contexts,
        "unique_unseen_contexts": len(unique_unseen_contexts),
        "unseen_context_pct": unseen_context_pct,
        "unseen_events": unseen_events,
        "unseen_event_pct": unseen_event_pct,
    }


print("=" * 100)
print("TRIGRAM SPARSITY DIAGNOSTICS ON TEST SET")
print("=" * 100)

trigram_diag = trigram_sparsity_diagnostics(trigram_lm, test_sents_unk)

print(f"Unique bigrams learned:          {len(bigram_lm.bigram_counts):,}")
print(f"Unique trigrams learned:         {len(trigram_lm.trigram_counts):,}")
print(f"Test trigram positions:          {trigram_diag['total_positions']:,}")
print(f"Unseen trigram contexts:         {trigram_diag['unseen_contexts']:,} ({trigram_diag['unseen_context_pct']:.2f}%)")
print(f"Unique unseen trigram contexts:  {trigram_diag['unique_unseen_contexts']:,}")
print(f"Unseen trigram events:           {trigram_diag['unseen_events']:,} ({trigram_diag['unseen_event_pct']:.2f}%)")
print("\nInterpretation:")
print("  High unseen-context or unseen-event percentages indicate trigram sparsity.")
print("  In that case, worse trigram perplexity is expected with plain add-alpha smoothing.")
print("=" * 100)
```

## Sentence Completion / Autocomplete with Beam Search

For sentence autocomplete, this notebook uses **beam search decoding** only.

Beam search keeps several candidate completions at each step instead of committing immediately to a single next word. With `beam_width = 3`, the decoder expands all current candidate sequences, scores them using cumulative log2 probabilities, and keeps only the three best sequences.

The decoder preserves the existing preprocessing, vocabulary, smoothing, and START/END token handling. It also never generates `*UNK*`, so the output remains readable.

```python
def normalize_prefix(prefix):
    """Apply the same preprocessing and OOV replacement used for corpus sentences."""
    tokens = preprocess_sentence(prefix.split())
    return replace_oov(tokens, vocabulary)


def generate_bigram_beam(model, prefix, beam_width=3, max_len=20):
    """Generate text from a bigram model using beam search."""
    beams = [(normalize_prefix(prefix), 0.0, False)]

    while True:
        candidates = []

        for tokens, score, finished in beams:
            if finished or len(tokens) >= max_len:
                candidates.append((tokens, score, finished))
                continue

            context = tokens[-1] if tokens else START

            for word in model.vocabulary:
                if word == UNK:
                    continue

                prob = model.probability(context, word)
                new_tokens = tokens + [word]
                new_score = score + math.log2(prob)
                candidates.append((new_tokens, new_score, word == END))

        candidates.sort(key=lambda item: item[1], reverse=True)
        beams = candidates[:beam_width]

        if beams[0][2] or all(finished or len(tokens) >= max_len for tokens, _, finished in beams):
            break

    return " ".join(beams[0][0])


def generate_trigram_beam(model, prefix, beam_width=3, max_len=20):
    """Generate text from a trigram model using beam search."""
    beams = [(normalize_prefix(prefix), 0.0, False)]

    while True:
        candidates = []

        for tokens, score, finished in beams:
            if finished or len(tokens) >= max_len:
                candidates.append((tokens, score, finished))
                continue

            context_tokens = [START1, START2] + tokens
            context = (context_tokens[-2], context_tokens[-1])

            for word in model.vocabulary:
                if word == UNK:
                    continue

                prob = model.probability(context, word)
                new_tokens = tokens + [word]
                new_score = score + math.log2(prob)
                candidates.append((new_tokens, new_score, word == END))

        candidates.sort(key=lambda item: item[1], reverse=True)
        beams = candidates[:beam_width]

        if beams[0][2] or all(finished or len(tokens) >= max_len for tokens, _, finished in beams):
            break

    return " ".join(beams[0][0])


print("Beam search autocomplete functions defined")
print("beam_width = 3")
print("max_len = 20")
print("Scoring: cumulative log2 probabilities")
print("UNK generation: disabled")
```

```python
print("=" * 120)
print("BEAM SEARCH AUTOCOMPLETE DEMONSTRATIONS")
print("=" * 120)

prompts = [
    "i would like to",
    "the president of",
    "he was",
    "in the",
    "the united states",
]

results = []

for prefix in prompts:
    bigram_completion = generate_bigram_beam(bigram_lm, prefix, beam_width=3, max_len=20)
    trigram_completion = generate_trigram_beam(trigram_lm, prefix, beam_width=3, max_len=20)

    print(f"\nPrompt:\n{prefix!r}\n")
    print("Bigram Beam Search:")
    print(bigram_completion)
    print("\nTrigram Beam Search:")
    print(trigram_completion)

    if len(trigram_completion.split()) >= len(bigram_completion.split()) and END not in trigram_completion.split()[:3]:
        comment = "The trigram beam completion appears more fluent here because it uses a two-word context."
    else:
        comment = "The trigram beam completion is not clearly more fluent here, although it still uses richer context than the bigram model."
    print("\nComment:")
    print(comment)
    print("-" * 120)

    results.append({
        "prefix": prefix,
        "bigram": bigram_completion,
        "trigram": trigram_completion,
    })

print("\nVerification:")
print("  UNK in bigram beam outputs:", any(UNK in item["bigram"].split() for item in results))
print("  UNK in trigram beam outputs:", any(UNK in item["trigram"].split() for item in results))
print("=" * 120)
```

## Interesting Cases

The examples above are enough for a short qualitative discussion. The goal here is simply to note where the models behave well, where they are awkward or repetitive, and whether the trigram beam completion looks more coherent than the bigram beam completion.

```python
print("=" * 80)
print("INTERESTING CASES - SHORT SUMMARY")
print("=" * 80)

for item in results:
    bigram = item["bigram"]
    trigram = item["trigram"]

    bigram_repeats = len(bigram.split()) != len(set(bigram.split()))
    trigram_repeats = len(trigram.split()) != len(set(trigram.split()))

    print(f"\nPrompt: {item['prefix']!r}")
    print(f"Bigram beam:  {bigram}")
    print(f"Trigram beam: {trigram}")

    if not trigram_repeats and len(trigram.split()) >= len(bigram.split()):
        print("Comment: trigram beam is the stronger/more fluent completion here.")
    elif bigram_repeats or trigram_repeats:
        print("Comment: at least one output shows repetition, a common n-gram limitation.")
    else:
        print("Comment: both outputs are usable, but fluency should be judged qualitatively.")

print("\nOverall: trigram beam search usually has the advantage because it uses a two-word context, but both models can still suffer from sparsity and repetition.")
```

- Perplexity gives a compact held-out comparison, while beam-search autocomplete examples show qualitative behavior.
- UNK replacement keeps evaluation well-defined for rare and unseen words.
- Beam search keeps the top three candidate continuations at each step using cumulative log2 probability.

### Sparsity

N-gram models become sparse quickly as `n` increases. A trigram context is more informative than a bigram context, but many possible trigram combinations never appear in training. Add-alpha smoothing prevents zero probabilities, but it cannot fully solve the lack of evidence for rare contexts.

### Why Bigram Can Sometimes Look Better

Although trigram models usually have the advantage of more context, a bigram can sometimes produce better-looking results or lower perplexity. This can happen because trigram contexts are much sparser, especially in a specialized corpus such as Reuters. The bigram model has less context, but its estimates are often based on more frequent events.

## Conclusion

The trigram model is generally stronger than the bigram model because it uses a richer two-word context, which often improves fluency and autocomplete quality. The development-set tuning step also makes the final test evaluation more principled because alpha is selected before looking at test performance.

The main weakness of n-gram language models is sparsity. As the context gets longer, many valid word sequences are missing from the training data, so smoothing becomes necessary. However, smoothing only redistributes probability mass; it does not give the model true semantic understanding.

Overall, n-gram models are useful, interpretable baselines for language modeling, but they are limited by fixed context length, poor handling of rare words, lack of semantic similarity, and inability to model long-range dependencies.
